```python
import pandas as pd

# Odor group assignments
GROUP_A_ODORS = {2, 3, 6, 10, 11, 12}
GROUP_B_ODORS = {1, 4, 5, 7, 8, 9}

def compute_participant_level_stats(phase1_path, phase2_path):
    # Load data
    df1 = pd.read_excel(phase1_path, skiprows=1)
    df2 = pd.read_excel(phase2_path, skiprows=1)

    df1.columns = df1.columns.str.strip()
    df2.columns = df2.columns.str.strip()

    participant_results = []

    for i in range(len(df2)):
        pid = int(str(df2.iloc[i, 0]).strip())  # Participant ID
        group = str(df2.iloc[i, 1]).strip().upper()
        responses = df2.iloc[i, 2::2].reset_index(drop=True)  # Recognition answers
        confidence = df2.iloc[i, 3::2].reset_index(drop=True)  # Confidence ratings (optional)

        # Determine known odors
        if group == 'A':
            known_odors = GROUP_A_ODORS
        elif group == 'B':
            known_odors = GROUP_B_ODORS
        else:
            raise ValueError(f"Unknown group for participant {pid}: {group}")

        # Initialize counters and trackers
        hits = misses = false_alarms = correct_rejections = 0
        hit_odors = []
        miss_odors = []
        fa_odors = []
        cr_odors = []

        for odor_idx, answer in enumerate(responses, start=1):
            answer = str(answer).strip().lower()
            is_old = odor_idx in known_odors

            if is_old:
                if answer == "yes":
                    hits += 1
                    hit_odors.append(odor_idx)
                elif answer == "no":
                    misses += 1
                    miss_odors.append(odor_idx)
            else:
                if answer == "yes":
                    false_alarms += 1
                    fa_odors.append(odor_idx)
                elif answer == "no":
                    correct_rejections += 1
                    cr_odors.append(odor_idx)

        # Compute rates
        hit_rate = hits / (hits + misses) if (hits + misses) > 0 else None
        false_alarm_rate = false_alarms / (false_alarms + correct_rejections) if (false_alarms + correct_rejections) > 0 else None

        participant_results.append({
            "Participant": pid,
            "Group": group,
            "Hits": hits,
            "Hit Odors": hit_odors,
            "Misses": misses,
            "Miss Odors": miss_odors,
            "False Alarms": false_alarms,
            "False Alarm Odors": fa_odors,
            "Correct Rejections": correct_rejections,
            "Correct Rejection Odors": cr_odors,
            "Hit Rate": hit_rate,
            "False Alarm Rate": false_alarm_rate
        })

    return pd.DataFrame(participant_results)


phase1_path = "/Users/liuwenjin/Downloads/odorants rating - phase 1.xlsx"
phase2_path = "/Users/liuwenjin/Downloads/odorants rating -- Phase2_June 29, 2025_20.12.xlsx"

participant_df = compute_participant_level_stats(phase1_path, phase2_path)
participant_df
```

```python
import pandas as pd
from collections import defaultdict

# Odor group assignments
GROUP_A_ODORS = {2, 3, 6, 10, 11, 12}
GROUP_B_ODORS = {1, 4, 5, 7, 8, 9}

# Map confidence to numerical scores (based on likely scale)
CONFIDENCE_MAP = {
    "not at all confident": 1,
    "slightly confident": 2,
    "somewhat confident": 3,
    "moderately confident": 4,
    "very confident": 5,
    "extremely confident": 6,
    "absolutely certain": 7
}

def compute_order_level_summary(phase1_path, phase2_path):
    # Load data
    df1 = pd.read_excel(phase1_path, skiprows=1)
    df2 = pd.read_excel(phase2_path, skiprows=1)
    df1.columns = df1.columns.str.strip()
    df2.columns = df2.columns.str.strip()

    # Initialize trackers
    hit_counts = defaultdict(int)
    miss_counts = defaultdict(int)
    fa_counts = defaultdict(int)
    cr_counts = defaultdict(int)
    pleasantness_scores = defaultdict(list)
    intensity_scores = defaultdict(list)
    confidence_scores = defaultdict(list)

    for i in range(len(df2)):
        pid = int(str(df2.iloc[i, 0]).strip())
        group = str(df2.iloc[i, 1]).strip().upper()
        responses = df2.iloc[i, 2::2].reset_index(drop=True)  # Yes/No
        confidence = df2.iloc[i, 3::2].reset_index(drop=True)  # Confidence

        # Determine known odors for Phase 1
        known_odors = GROUP_A_ODORS if group == 'A' else GROUP_B_ODORS

        for odor_idx in range(1, 13):
            recog = str(responses[odor_idx - 1]).strip().lower()
            conf = str(confidence[odor_idx - 1]).strip().lower()
            conf_score = CONFIDENCE_MAP.get(conf, None)
            if conf_score:
                confidence_scores[odor_idx].append(conf_score)

            if odor_idx in known_odors:
                if recog == "yes":
                    hit_counts[odor_idx] += 1
                elif recog == "no":
                    miss_counts[odor_idx] += 1
            else:
                if recog == "yes":
                    fa_counts[odor_idx] += 1
                elif recog == "no":
                    cr_counts[odor_idx] += 1

    # Process Phase 1 for pleasantness and intensity
    for i in range(len(df1)):
        group = str(df1.iloc[i, 1]).strip().upper()
        known_odors = GROUP_A_ODORS if group == 'A' else GROUP_B_ODORS
        responses = df1.iloc[i]

        for local_idx, odor_id in enumerate(sorted(known_odors)):
            intensity_col = 2 + local_idx * 2
            pleasantness_col = intensity_col + 1

            try:
                intensity = float(responses[intensity_col])
                pleasant = str(responses[pleasantness_col]).strip().lower()

                # Convert pleasantness labels to numeric values (assume ordinal)
                pleasant_score_map = {
                    "extremely unpleasant": 1,
                    "unpleasant": 2,
                    "moderately unpleasant": 3,
                    "mildly unpleasant": 4,
                    "neutral": 5,
                    "mildly pleasant": 6,
                    "moderately pleasant": 7,
                    "pleasant": 8,
                    "extremely pleasant": 9
                }

                pleasant_score = pleasant_score_map.get(pleasant, None)

                if pleasant_score:
                    pleasantness_scores[odor_id].append(pleasant_score)
                intensity_scores[odor_id].append(intensity)
            except:
                continue  # Handle missing or malformed input

    # Assemble final summary
    summary = []
    for odor in range(1, 13):
        hits = hit_counts[odor]
        misses = miss_counts[odor]
        fas = fa_counts[odor]
        crs = cr_counts[odor]

        hit_rate = hits / (hits + misses) if (hits + misses) > 0 else None
        fa_rate = fas / (fas + crs) if (fas + crs) > 0 else None

        pleasant_mean = sum(pleasantness_scores[odor]) / len(pleasantness_scores[odor]) if pleasantness_scores[odor] else None
        intensity_mean = sum(intensity_scores[odor]) / len(intensity_scores[odor]) if intensity_scores[odor] else None
        confidence_mean = sum(confidence_scores[odor]) / len(confidence_scores[odor]) if confidence_scores[odor] else None

        summary.append({
            "Odor": odor,
            "Hit Rate": hit_rate,
            "False Alarm Rate": fa_rate,
            "Perceived Pleasantness": pleasant_mean,
            "Intensity": intensity_mean,
            "Confidence": confidence_mean
        })

    return pd.DataFrame(summary)
```

```python
phase1_path = "/Users/liuwenjin/Downloads/odorants rating - phase 1.xlsx"
phase2_path = "/Users/liuwenjin/Downloads/odorants rating -- Phase2_June 29, 2025_20.12.xlsx"


order_df = compute_order_level_summary(phase1_path, phase2_path)
order_df['Correct Recognition'] = order_df['Hit Rate'] - order_df['False Alarm Rate']
order_df
```

```python
# Q: Are certain odors consistently more memorable across young adults, or is odor memorability highly individual-specific?


# GROUP DEFINITION
GROUP_A_ODORS = {2, 3, 6, 10, 11, 12}
GROUP_B_ODORS = {1, 4, 5, 7, 8, 9}

# Assume participant_df is already loaded and contains: 'Participant', 'Group', 'Hit Odors', 'Miss Odors'

# ✅ Step 1: Build odor × participant memory matrix, enforcing group exposure rule
odor_hit_matrix = pd.DataFrame(index=participant_df['Participant'], columns=range(1, 13), dtype=float)

for _, row in participant_df.iterrows():
    pid = row['Participant']
    group = row['Group']
    known_odors = GROUP_A_ODORS if group == 'A' else GROUP_B_ODORS

    for odor in row['Hit Odors']:
        if odor in known_odors:
            odor_hit_matrix.at[pid, odor] = 1
    for odor in row['Miss Odors']:
        if odor in known_odors:
            odor_hit_matrix.at[pid, odor] = 0
    # All other cells remain NaN (not exposed)

# ✅ Step 2: Heatmap of memory matrix (Hit = 1, Miss = 0, NaN = didn't smell)
plt.figure(figsize=(10, 6))
sns.heatmap(
    odor_hit_matrix,
    cmap='coolwarm',
    cbar_kws={'label': 'Memory (1 = Hit, 0 = Miss)'},
    linewidths=0.5,
    linecolor='lightgray'
)
plt.title("Participant × Odor Memorability (Hit = 1, Miss = 0, NaN = Not Smelled)")
plt.xlabel("Odor Index")
plt.ylabel("Participant")
plt.tight_layout()
plt.show()

# ✅ Step 3: Bar plot — How many participants remembered each odor
hit_counts = odor_hit_matrix.sum(skipna=True)
n_smelled = odor_hit_matrix.notna().sum()
hit_rate_per_odor = (hit_counts / n_smelled).sort_index()

plt.figure(figsize=(8, 4))
hit_rate_per_odor.plot(kind='bar', color='skyblue', edgecolor='black')
plt.title("Hit Rate per Odor (Only for Participants Who Smelled It)")
plt.xlabel("Odor Index")
plt.ylabel("Hit Rate")
plt.ylim(0, 1)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()

```

```python


# Step 1: Descriptive statistics
print("\n📊 Descriptive Stats on Pleasantness & Confidence")
print(order_df[['Perceived Pleasantness', 'Confidence']].describe().round(2))

# Step 2: Correlations
print("\n📈 Correlation Between Pleasantness and Hit Rate / Confidence")
df_corr = order_df.dropna(subset=['Perceived Pleasantness', 'Hit Rate', 'Confidence'])

# Pearson & Spearman
pearson_hit, p_hit = pearsonr(df_corr['Perceived Pleasantness'], df_corr['Hit Rate'])
spearman_hit, sp_hit = spearmanr(df_corr['Perceived Pleasantness'], df_corr['Hit Rate'])

pearson_conf, p_conf = pearsonr(df_corr['Perceived Pleasantness'], df_corr['Confidence'])
spearman_conf, sp_conf = spearmanr(df_corr['Perceived Pleasantness'], df_corr['Confidence'])

print(f"🔹 Pleasantness ↔ Hit Rate:     Pearson r = {pearson_hit:.3f}, p = {p_hit:.4f} | Spearman ρ = {spearman_hit:.3f}, p = {sp_hit:.4f}")
print(f"🔹 Pleasantness ↔ Confidence:  Pearson r = {pearson_conf:.3f}, p = {p_conf:.4f} | Spearman ρ = {spearman_conf:.3f}, p = {sp_conf:.4f}")

# Step 3: Visualization - Pleasantness vs Hit Rate
plt.figure(figsize=(6, 4))
sns.regplot(data=df_corr, x='Perceived Pleasantness', y='Hit Rate', ci=95, scatter_kws={"s": 60}, line_kws={"color": "red"})
plt.title("Pleasantness vs Recognition Accuracy (Hit Rate)")
plt.xlabel("Perceived Pleasantness")
plt.ylabel("Hit Rate")
plt.grid(True, linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()

# Step 4: Visualization - Pleasantness vs Confidence
plt.figure(figsize=(6, 4))
sns.regplot(data=df_corr, x='Perceived Pleasantness', y='Confidence', ci=95, scatter_kws={"s": 60}, line_kws={"color": "red"})
plt.title("Pleasantness vs Confidence")
plt.xlabel("Perceived Pleasantness")
plt.ylabel("Confidence")
plt.grid(True, linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
```

```python
# Step 3: ANOVA with meaningful pleasantness labels
print("\n🧪 ANOVA: Recognition Accuracy Across Each Pleasantness Level")

# Copy and convert score to string labels (1–9)
df_anova = df_corr.copy()
df_anova['Pleasantness_Label'] = df_anova['Perceived Pleasantness'].astype(int).astype(str)

# Run ANOVA using specific score levels as categories
model = ols('Q("Hit Rate") ~ C(Pleasantness_Label)', data=df_anova).fit()
anova_table = sm.stats.anova_lm(model, typ=2)
print(anova_table)

# 📊 Visualization: Boxplot by each specific pleasantness score
plt.figure(figsize=(9, 5))
sns.boxplot(data=df_anova, x='Pleasantness_Label', y='Hit Rate', palette='coolwarm')
sns.stripplot(data=df_anova, x='Pleasantness_Label', y='Hit Rate', color='black', size=6, jitter=True, alpha=0.5)

plt.title("Hit Rate (Recognition Accuracy) by Each Pleasantness Level")
plt.xlabel("Pleasantness Score (1 = Extremely Unpleasant → 9 = Extremely Pleasant)")
plt.ylabel("Hit Rate")
plt.ylim(0, 1)
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()

```
