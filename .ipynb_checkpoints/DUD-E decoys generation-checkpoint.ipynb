```python
# Check overlap between active ligands.smi files

def find_overlapping_ligands(directory):
    smi_files = [f for f in os.listdir(directory) if f.endswith('.smi')]
    ligands_map = defaultdict(set)  # ligand_id -> set of file names

    for smi_file in smi_files:
        file_path = os.path.join(directory, smi_file)
        with open(file_path, 'r') as f:
            for line in f:
                if line.strip():
                    try:
                        smiles, ligand_id = line.strip().split('\t')
                        ligands_map[ligand_id].add(smi_file)
                    except ValueError:
                        print(f"⚠️ Malformed line in {smi_file}: {line.strip()}")

    # Find duplicates: ligand IDs that appear in >1 file
    overlapping = {ligand: files for ligand, files in ligands_map.items() if len(files) > 1}

    if overlapping:
        print("🔍 Overlapping ligand IDs found across files:")
        for ligand_id, files in overlapping.items():
            print(f"{ligand_id} appears in: {sorted(files)}")
    else:
        print("✅ No overlapping ligand IDs found across the .smi files.")

    return overlapping

find_overlapping_ligands('/Users/liuwenjin/Downloads/')
```

```python
# Verify that DUD-E generated docoys cover all active ligands

def verify_dude_decoys(input_active_smi_path, dude_output_dir, save_name):
    # Step 1: Load input actives
    with open(input_active_smi_path, 'r') as f:
        input_actives = {line.strip().split('\t')[1] for line in f if line.strip()}

    print(f"📄 Number of ligands in input .smi file: {len(input_actives)}")

    # Step 2: Locate and count .picked files
    decoy_dir = os.path.join(dude_output_dir, 'decoys')
    picked_files = sorted(f for f in os.listdir(decoy_dir) if f.endswith('.picked'))

    print(f"📂 Number of .picked files in DUD-E decoys folder: {len(picked_files)}")
    if len(input_actives) != len(picked_files):
            print(f"⚠️ active.smi number and.picked file number mismatch")

    all_df = []

    for file in picked_files:
        file_path = os.path.join(decoy_dir, file)
        with open(file_path, 'r') as f:
            lines = f.readlines()

        # Process first line: remove first token (e.g. "ligand")
        first_tokens = lines[0].strip().split('\t')
        if len(first_tokens) != 4 or first_tokens[0].lower() != "ligand":
            print(f"⚠️ Unexpected format in first line of {file}")
            continue

        first_row = first_tokens[1:]  # skip 'ligand'

        # Optional: skip if not in input actives
        if first_row[1] not in input_actives:
            continue

        # Process remaining lines (already correct format)
        rest_rows = [line.strip().split('\t') for line in lines[1:] if line.strip()]

        # Combine all rows
        all_rows = [first_row] + rest_rows
        df = pd.DataFrame(all_rows, columns=['smiles', 'id', 'internal_id'])

        # Add label column
        df['label'] = df['id'].apply(lambda x: 'active' if x in input_actives else 'decoy')

        # Check counts
        num_actives = (df['label'] == 'active').sum()
        num_decoys = (df['label'] == 'decoy').sum()
        print(f"{file}: ✅ {num_actives} active, ✅ {num_decoys} decoys")
        if num_actives != 1 or num_decoys != 50:
            print(f"⚠️ Error: Unexpected count in {file}")

        all_df.append(df)

    # Step 3: Combine and check completeness
    combined_df = pd.concat(all_df, ignore_index=True)
    active_ids_in_combined = set(combined_df[combined_df['label'] == 'active']['id'])
    missing_actives = input_actives - active_ids_in_combined

    if missing_actives:
        print(f"❌ Missing actives: {missing_actives}")
    else:
        print("✅ All input actives are present in the DUD-E output")

    # Step 4: Save to file
    save_path = os.path.join(dude_output_dir, f"{save_name}.smi")
    combined_df[['smiles', 'id', 'label']].to_csv(save_path, sep='\t', index=False, header=False)
    print(f"✅ Combined SMILES saved to: {save_path}")

    return combined_df
```

```python
import os

def combine_all_dude_decoys(root_dir):
    combined_lines = []
    subdirs = [d for d in os.listdir(root_dir)
               if os.path.isdir(os.path.join(root_dir, d)) and d.startswith("DUDE-pAct")]

    has_pAct5 = any("pAct5" in d for d in subdirs)
    has_pAct6 = any("pAct6" in d for d in subdirs)

    # Decide output filename
    if has_pAct6 and not has_pAct5:
        output_name = "DUDE-pAct6-Active-Decoys.smi"
    else:
        output_name = "DUDE-pAct5-Active-Decoys.smi"

    # Collect SMILES lines
    for subdir in subdirs:
        smi_path = os.path.join(root_dir, subdir, 'DUDE-pAct-Active-Decoys.smi')
        if os.path.exists(smi_path):
            with open(smi_path, 'r') as f:
                lines = [line.strip() for line in f if line.strip()]
                combined_lines.extend(lines)
                print(f"✅ Added {len(lines)} lines from {smi_path}")
        else:
            print(f"⚠️ File not found: {smi_path}")

    # Save to output file
    output_path = os.path.join(root_dir, output_name)
    with open(output_path, 'w') as out_f:
        out_f.write('\n'.join(combined_lines) + '\n')

    print(f"\n✅ Combined total of {len(combined_lines)} entries written to {output_path}")
    return output_path

combine_all_dude_decoys('/Users/liuwenjin/Downloads/')
```
